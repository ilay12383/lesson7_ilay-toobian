```python

def multiply(a, b):
    return a * b

def is_even(number):
    return number % 2 == 0

result1 = multiply(4, 5)
result2 = multiply(7, 3)

print("4 * 5 =", result1)
print("7 * 3 =", result2)

print("Is 10 even?", is_even(10))
print("Is 7 even?", is_even(7))
```

```python

def celsius_to_fahrenheit(celsius):
    fahrenheit = celsius * 9/5 + 32
    return fahrenheit

print("0°C =", celsius_to_fahrenheit(0), "°F")
print("25°C =", celsius_to_fahrenheit(25), "°F")
print("100°C =", celsius_to_fahrenheit(100), "°F")
```

```python

def get_grade(score):
    if 90 <= score <= 100:
        return "A"
    elif 80 <= score <= 89:
        return "B"
    elif 70 <= score <= 79:
        return "C"
    elif 60 <= score <= 69:
        return "D"
    else:
        return "F"

test_scores = [95, 82, 76, 64, 50]

for score in test_scores:
    print(f"Score: {score} -> Grade: {get_grade(score)}")
```

```python
def analyze_list(numbers):
    total = sum(numbers)
    count = len(numbers)
    average = total / count if count > 0 else 0
    return total, average, count

test_scores = [95, 82, 76, 64, 50]

total, avg, count = analyze_list(test_scores)
print(f"Total: {total}, Average: {avg}, Count: {count}")
```

```python

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Division by zero!"
    return a / b

def calculator():
    print("Mini Calculator")
    print("Select operation:")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")

    choice = input("Enter choice (1/2/3/4): ")

    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
    except ValueError:
        print("Invalid input! Please enter numbers only.")
        return

    if choice == "1":
        result = add(num1, num2)
        operation = "+"
    elif choice == "2":
        result = subtract(num1, num2)
        operation = "-"
    elif choice == "3":
        result = multiply(num1, num2)
        operation = "*"
    elif choice == "4":
        result = divide(num1, num2)
        operation = "/"
    else:
        print("Invalid choice!")
        return

    print(f"{num1} {operation} {num2} = {result}")


calculator()
```

```python

def power(base, exponent):
    return base ** exponent

def factorial(n):
    if n < 0:
        return "Error: factorial not defined for negative numbers"
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print("Power 2^5 =", power(2, 5))
print("Factorial 5! =", factorial(5))
print("Factorial 0! =", factorial(0))



def count_vowels(text):
    vowels = "aeiouAEIOU"2
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count

def reverse_string(text):
    return text[::-1]

#
test_text1 = "Hello World"
test_text2 = "Python is fun!"

print(f"Vowels in '{test_text1}':", count_vowels(test_text1))
print(f"Reversed '{test_text2}':", reverse_string(test_text2))



def is_valid_email(email):
    return "@" in email and "." in email

def is_strong_password(password):
    return len(password) >= 8

emails = ["test@example.com", "invalidemail", "hello@world"]
passwords = ["mypassword", "123", "StrongPass123"]

for email in emails:
    print(f"Email '{email}' valid?", is_valid_email(email))

for pwd in passwords:
    print(f"Password '{pwd}' strong?", is_strong_password(pwd))
```
